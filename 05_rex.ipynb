```python
import pandas as pd
import os
from causalexplain.estimators.rex.rex import Rex

# Verified path from user
file_path = '/content/drive/MyDrive/ml/CORRECTEDv3/all_cities_combined_v3.parquet'

if not os.path.exists(file_path):
    print(f"Error: File not found at {file_path}.")
else:
    # Load the parquet dataset
    data = pd.read_parquet(file_path)
    print("Dataset loaded successfully.")

    # Based on previous output, the actual column name is 'eta_mins'
    target = 'eta_mins'

    if target not in data.columns:
        print(f"Error: Column '{target}' not found. Available columns: {data.columns.tolist()}")
    else:
        # Rex works best with numeric features; filtering out non-numeric columns
        numeric_data = data.select_dtypes(include=['number'])

        # Initialize Rex
        rex_model = Rex(numeric_data, target=target)

        # Prepare features X by dropping the target column
        X = numeric_data.drop(columns=[target])

        print(f"\nRunning Rex algorithm on target: {target}...")
        # Pass X to the fit method as required by the library
        rex_model.fit(X)

        print("\nRex Algorithm Results:")
        display(rex_model.explain())
```
